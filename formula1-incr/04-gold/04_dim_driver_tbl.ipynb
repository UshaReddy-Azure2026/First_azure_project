```python
driver_df=spark.read.table("formula1.silver.driver")
```

```python
driver_na_df=spark.read.table("formula1.gold.ref_nationality_region")
```

```python
driver_final_df=driver_df.join(driver_na_df,driver_df.nationality==driver_na_df.nationality,"left").select(
    driver_df.driver_id,
    driver_df.driver_name,
    driver_df.date_of_birth,
    driver_df.nationality,
    driver_na_df.region
)
```

```python
driver_final_df.write.format("delta").mode("overwrite").saveAsTable("formula1.gold.dim_drivers")
```
