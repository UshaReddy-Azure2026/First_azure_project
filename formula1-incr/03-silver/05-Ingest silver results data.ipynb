```python
bronze_table=f"{catalog_name}.{bronze_schema}.results"
silver_table=f"{catalog_name}.{silver_schema}.results"
```

```python
from pyspark.sql import functions as F
```

```python
results_df=spark.read.table(bronze_table)
```

```python
results_select_df=results_df.select(F.col("season"),
                F.col("round"),
                 F.col("constructorId").alias("constructor_id"),
                 F.col("driverId").alias("driver_id"),
                 F.col("date").alias("race_date"),
                 F.initcap(F.col("raceName")).alias("race_name"),
                 F.col("grid").alias("grid_position"),
                 F.col("laps").alias("completed_laps"),
                 F.col("number").alias("car_number"),
                 F.col("points"),
                 F.col("position"),
                 F.col("positionText").alias("finish_position_text"),
                 F.col("status"),
                 F.col("ingestion_timestamp"),
                 F.col("source_file"))
```

```python
results_select_rem_df=results_select_df.dropDuplicates()
```

```python
results_select_fin_df=results_select_rem_df.filter("season is not null or round is not null or constructor_id is not null or driver_id is not null")
```

```python
results_select_fin_df.write.format("delta").mode("overwrite").saveAsTable(silver_table)
```
