```python

from pyspark.sql import functions as F

def functions_helpers (df):
  return(df.withColumn ('ingestion_timestamp', F.current_timestamp())
         .withColumn ('source_file', F.col('_metadata.file_path')))
```

```python
def circute_tabel_fuc(
    input_df,
    target_table,
batch_id):
    circutes_final_df1 = input_df.withColumn("batch_id", F.lit(batch_id))
    circutes_final_df1.write.format('delta').partitionBy("batch_id").option("replaceWhere", f"batch_id='{batch_id}'").mode('overwrite').saveAsTable(target_table)
```
