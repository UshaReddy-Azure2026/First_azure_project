```python
results_df=spark.read.table("formula1.silver.results")
```

```python
sprint_df=spark.read.table("formula1.silver.sprints")
```

```python
from pyspark.sql import functions as F
```

```python
results_df1=results_df.withColumn("sessio_type",F.lit("RACE"))
```

```python
results_df1=sprint_df.withColumn("sessio_type",F.lit("SPRINT"))
```

```python
results_union_df=results_df1.unionByName(results_df1)
```

```python
results_union_df1=(results_union_df
.withColumn("is_win",F.col("finish_position")==1)
.withColumn ("is_podium",F.col("finish_position").between(1,3)))
#withColumn("has_points",F.col("points")>0))
```

```python
results_union_df1.display()
```

```python
results_union_df1.write.format("delta").mode("overwrite").saveAsTable("formula1.gold.dim_results")
```
