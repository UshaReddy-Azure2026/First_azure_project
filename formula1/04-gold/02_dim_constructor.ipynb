```python
constr_df=spark.read.table("formula1.silver.constructor")
```

```python
constr_na_df=spark.read.table("formula1.gold.ref_nationality_region")
```

```python
constr_join_df=constr_df.join(constr_na_df,constr_na_df.nationality==constr_df.nationality,"left").select(
constr_df.constructor_id,
constr_df.constructor_name,
constr_na_df.nationality,
constr_na_df.region)
```

```python
constr_join_df.write.format("delta").mode("overwrite").saveAsTable("formula1.gold.dim_constructors")
```
