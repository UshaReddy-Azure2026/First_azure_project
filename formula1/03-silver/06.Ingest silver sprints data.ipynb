```python
bronze_table =f"{catalog_name}.{bronze_schema}.sprints"
silver_table =f"{catalog_name}.{silver_schema}.sprints"
```

```python
silver_sprints_df=spark.read.table(bronze_table)
```

```python
silver_sprints_select_df=silver_sprints_df.select("date","raceName","round","season","constructorId",'driverId','grid','laps',
                         'number','position','positionText','status','ingestion_timestamp','source_file')
```

```python
silver_sprints_final_df=silver_sprints_select_df.withColumnsRenamed({"constructorId": "constructor_id","driverId": "driver_id","raceName":"race_name","positionText":"finish_position_text","date":"race_date","grid":"grid_position","laps":"completed_laps","number":"car_number","position":"finish_position"})
```

```python
from pyspark.sql import functions as F
```

```python
silver_sprints_final1_df=silver_sprints_final_df.withColumn("race_name",F.initcap(F.col("race_name")))
```

```python
silver_sprints_final1_df.write.format("delta").mode("overwrite").saveAsTable(silver_table)
```
