```python
bronze_table=f"{catalog_name}.{bronze_schema}.races"
silver_table=f"{catalog_name}.{silver_schema}.races"
```

```python
races_df=spark.read.table(bronze_table)
```

```python
from pyspark.sql import functions as sf
```

```python
races_final_df=races_df.select(sf.col("season")
                               ,sf.col("round"),
                               sf.col("raceName").alias("race_name"),
                               sf.col('date').alias("race_dat"),
                               sf.col('circuitId').alias("circuit_id"),
                               sf.col('ingestion_timestamp'),
                               sf.col('source_file')
                               )
```

```python
races_final1_df=races_final_df.distinct()
```

```python
races_final2_df=races_final1_df.withColumn("race_name", sf.initcap(sf.col("race_name")))
```

```python
races_final2_df.write.format("delta").mode("overwrite").saveAsTable(silver_table)
```
