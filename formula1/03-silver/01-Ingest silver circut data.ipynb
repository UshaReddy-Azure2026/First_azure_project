```python
bronze_table=f"{catalog_name}.{bronze_schema}.circuts"
silver_table=f"{catalog_name}.{silver_schema}.circuts"
bronze_table
```

```python
silver_circut=spark.read.table(bronze_table)
```

```python
silver_select_circut=silver_circut.select(
"circuitId","circuitName","lat","long","locality","country","ingestion_timestamp","source_file"
)
```

```python
display(silver_select_circut)
```

```python
from pyspark.sql import functions as F
```

```python
silver_select_circut=silver_circut.select(
F.col("circuitId"),
F.col("circuitName"),
F.col("lat"),
F.col("long"),
F.col("locality"),
F.col("country"),
F.col("ingestion_timestamp"),
F.col("source_file")
)
```

```python
silver_reamen_circut_df=(silver_select_circut.withColumnsRenamed({"circuitId":"circuit_id",
"circuitName":"circuit_name",
"lat":"latitude",
"long":"longitude"})
)
```

```python
silver_circut_nonnull_df=(silver_reamen_circut_df.dropDuplicates())
```

```python
silver_nonull_df=silver_circut_nonnull_df.filter(F.col("circuit_id").isNotNull())
```

```python
cirrcuts_final_df=(silver_nonull_df.withColumn('circuit_name',F.initcap(F.col("circuit_name")))
.withColumn('locality',F.initcap(F.col("locality")))
)
```

```python
display(cirrcuts_final_df)
```

```python
cirrcuts_final_df.write.format("delta").mode("overwrite").saveAsTable(silver_table)
```
