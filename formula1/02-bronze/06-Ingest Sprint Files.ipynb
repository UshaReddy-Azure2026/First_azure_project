```python
source_full_path=f"{landing_folder_path}/sprints"
table_name=f"{catalog_name}.{bronze_schema}.sprints"
```

```python
#define the schema

from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType, DateType, TimestampType

sprint_schema = StructType([
  StructField("date", StringType(), True),
  StructField("raceName", StringType(), True),
  StructField("round", IntegerType(), True),
  StructField("season", IntegerType(), True),
  StructField("url", StringType(), True),
  StructField("constructorId",StringType(),True),
  StructField("driverId", StringType(), True),
  StructField("grid", IntegerType(), True),
  StructField("laps", IntegerType(), True),
  StructField("number", IntegerType(), True),
  StructField("position", IntegerType(), True),
  StructField("positionText", StringType(), True),
  StructField("status", StringType(), True)
])
```

```python
sprint_df=spark.read.format("json").option("multiline",True).schema(sprint_schema).load(source_full_path)

```

```python
sprint_final_df=functions_helpers(sprint_df)
```

```python
sprint_final_df.write.format("delta").mode("overwrite").saveAsTable(table_name)
```
