```python
source_file_path=f"{landing_folder_path}/drivers.json"
table_name=f"{catalog_name}.{bronze_schema}.driver"
```

```python
#define schema

from pyspark.sql.types import StructType, StructField, StringType,DateType
child_schema=StructType(
     [
         StructField('givenName',StringType(),True),
         StructField('familyName',StringType(),True)

     ]
 )
driver_schema=StructType(
     [
         StructField('driverId',StringType(), True),
         StructField('name',child_schema),
         StructField('dateOfBirth',DateType(),True),
         StructField('nationality',StringType(),True),
         StructField('url',StringType(),True)
     ]
 )
```

```python
driver_df=(spark.read.format("json")
.schema(driver_schema)
#.option("header",'true')
.load(source_file_path))
```

```python
display(driver_df)
```

```python
driver_final_df=functions_helpers(driver_df)
```

```python
display(driver_final_df)
```

```python
driver_final_df.write.format("delta").mode("overwrite").saveAsTable(table_name)
```
